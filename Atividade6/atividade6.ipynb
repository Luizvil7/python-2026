```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
tabela_2012 = pd.read_csv("H:\etapa3\Python\Atividade6\Tabela5-sem_emprego_2012.csv")
tabela_2026 = pd.read_csv("H:\etapa3\Python\Atividade6\Tabela5-sem_emprego_2026.csv")

tabela_2012 = tabela_2012.rename(
    columns={"Desocupados - mulheres (2012 T1)": "mulheres_2012"}
)
tabela_2026 = tabela_2026.rename(
    columns={"Desocupados - mulheres (2026 T1)": "mulheres_2026"}
)

tabela_2012

comp = tabela_2012.merge(tabela_2026, on=["Sigla", "Código", "Estado"], how ="inner")

ordem = comp.sort_values("mulheres_2026", ascending=False)["Estado"]

longo = comp.melt(
    id_vars=["Sigla", "Código", "Estado"],
    value_vars=["mulheres_2012", "mulheres_2026"],
    var_name="Ano",
    value_name="Participacao_mulheres",
)

longo["Ano"] = longo["Ano"].map (
    {"mulheres_2012": "2012 T1", "mulheres_2026": "2026 T1"}

)

fig, ax = plt.subplot(figsize=(10,10))

sns.barplot(
    data=longo,
    y="Estado",
    x="Participacao_mulheres",
    hue="Ano",
    order=ordem,
    ax=ax
)

ax.axvline(50, color="red", linestyle="--", linewidth=1) # MUDE A COR
ax.set_xlabel("Participação das mulheres entre as pessoas desocupadas (%)")
ax.set_ylabel("")
ax.set_title("Desocupação: participação feminina em 2012 T1 e 2026 T1")
ax.legend(title="Período")
fig.tight_layout()

plt.show()

```
